# 01 Data Loading

Load and explore the Nigerian Telecom/Mobile Money Dataset.

```python
import pandas as pd
from datasets import load_dataset
import os
```

```python
# Load dataset from Hugging Face
# Using the specified dataset for this project
print('Loading dataset from HuggingFace...')
dataset = load_dataset('electricsheepafrica/nigerian-telecom-fraudulent-activity-datasets', split='train')
df = dataset.to_pandas()
```

```python
print('\nDataset Info:')
print(df.info())
```

```python
print('\nFirst 5 rows:')
display(df.head())
```

```python
# Save to the local data directory so subsequent notebooks can load it easily
os.makedirs('../data', exist_ok=True)
df.to_csv('../data/raw_data.csv', index=False)
print('Dataset saved locally to ../data/raw_data.csv')
```
