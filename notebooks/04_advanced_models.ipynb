# 04 Advanced Models

Train Random Forest, XGBoost, and SVM models to handle class imbalance and complex nonlinear temporal relationships.

```python
import sys
import os
import pickle
sys.path.append(os.path.abspath('..'))

from src.train import train_and_predict
from src.evaluate import evaluate_model, print_results

with open('../data/processed/split_data.pkl', 'rb') as f:
    X_train, X_val, X_test, y_train, y_val, y_test = pickle.load(f)

# Load baseline results to append
with open('../outputs/results_baseline.pkl', 'rb') as f:
    results = pickle.load(f)
with open('../models/prob_dict_baseline.pkl', 'rb') as f:
    prob_dict = pickle.load(f)
```

```python
# 3. Random Forest
model_rf, y_pred_rf, y_prob_rf = train_and_predict('forest', X_train, y_train, X_val, y_val, X_test)

metric_rf = evaluate_model(y_test, y_pred_rf, y_prob_rf, model_name='Random Forest')
results.append(metric_rf)
prob_dict['Random Forest'] = y_prob_rf
```

```python
# 4. XGBoost
model_xgb, y_pred_xgb, y_prob_xgb = train_and_predict('xgboost', X_train, y_train, X_val, y_val, X_test)

metric_xgb = evaluate_model(y_test, y_pred_xgb, y_prob_xgb, model_name='XGBoost')
results.append(metric_xgb)
prob_dict['XGBoost'] = y_prob_xgb
```

```python
# 5. SVM
model_svm, y_pred_svm, y_prob_svm = train_and_predict('svm', X_train, y_train, X_val, y_val, X_test)

metric_svm = evaluate_model(y_test, y_pred_svm, y_prob_svm, model_name='SVM')
results.append(metric_svm)
prob_dict['SVM'] = y_prob_svm
```

```python
print_results(results)

with open('../models/prob_dict_adv.pkl', 'wb') as f:
    pickle.dump(prob_dict, f)
with open('../outputs/results_adv.pkl', 'wb') as f:
    pickle.dump(results, f)
```
