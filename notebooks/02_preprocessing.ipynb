# 02 Preprocessing

Clean the data, handle missing values, and perform a robust temporal train-test split (60-20-20).

```python
import sys
import os
import pandas as pd

# Add src directory to python path for imports
sys.path.append(os.path.abspath('..'))
from src.preprocess import load_and_preprocess_data, temporal_split_and_scale
```

```python
# Define column names based on dataset
time_col = 'detected_at'    # update if dataset has a different col name for time
target_col = 'status' # update if dataset has a different target name

# Load and clean the data. Generates time features (hour, day_of_week) and LabelEncodes categoricals.
df_clean = load_and_preprocess_data('../data/raw_data.csv', time_col=time_col, target_col=target_col)

display(df_clean.head())
```

```python
# Perform sequential/temporal split (60% train, 20% val, 20% test)
# We also apply StandardScaler (fit on train only!)
X_train, X_val, X_test, y_train, y_val, y_test = temporal_split_and_scale(df_clean, target_col=target_col)

# Save processed datasets for training notebooks
import pickle
os.makedirs('../data/processed', exist_ok=True)
with open('../data/processed/split_data.pkl', 'wb') as f:
    pickle.dump((X_train, X_val, X_test, y_train, y_val, y_test), f)

print('Preprocessing complete! Data saved locally in ../data/processed/')
```
