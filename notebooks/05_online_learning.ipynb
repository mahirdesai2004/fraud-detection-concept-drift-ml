# 05 Online Learning

Implement SGDClassifier for incremental learning. Online learning models adapt batches of data sequentially, which helps to naturally combat Concept Drift as it learns from newer patterns immediately without retaining all historic data.

```python
import sys
import os
import pickle
sys.path.append(os.path.abspath('..'))

from src.train import train_and_predict
from src.evaluate import evaluate_model, print_results

with open('../data/processed/split_data.pkl', 'rb') as f:
    X_train, X_val, X_test, y_train, y_val, y_test = pickle.load(f)

with open('../outputs/results_adv.pkl', 'rb') as f:
    results = pickle.load(f)
with open('../models/prob_dict_adv.pkl', 'rb') as f:
    prob_dict = pickle.load(f)
```

```python
# 6. SGD Classifier (Online Learning Simulation)
model_sgd, y_pred_sgd, y_prob_sgd = train_and_predict('sgd', X_train, y_train, X_val, y_val, X_test)

metric_sgd = evaluate_model(y_test, y_pred_sgd, y_prob_sgd, model_name='SGD (Online)')
results.append(metric_sgd)
prob_dict['SGD (Online)'] = y_prob_sgd
```

```python
print_results(results)

with open('../models/prob_dict_final.pkl', 'wb') as f:
    pickle.dump(prob_dict, f)
with open('../outputs/results_final.pkl', 'wb') as f:
    pickle.dump(results, f)
```
