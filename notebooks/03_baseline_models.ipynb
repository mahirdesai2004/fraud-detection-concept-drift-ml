# 03 Baseline Models

Train Logistic Regression and Decision Tree models to establish a baseline.

```python
import sys
import os
import pickle
sys.path.append(os.path.abspath('..'))

from src.train import train_and_predict
from src.evaluate import evaluate_model, print_results, save_results

with open('../data/processed/split_data.pkl', 'rb') as f:
    X_train, X_val, X_test, y_train, y_val, y_test = pickle.load(f)

results = []
prob_dict = {}  # for ROC/PR curves later
```

```python
# 1. Logistic Regression
model_log, y_pred_log, y_prob_log = train_and_predict('logistic', X_train, y_train, X_val, y_val, X_test)

metric_log = evaluate_model(y_test, y_pred_log, y_prob_log, model_name='Logistic Regression')
results.append(metric_log)
prob_dict['Logistic Regression'] = y_prob_log
```

```python
# 2. Decision Tree
model_tree, y_pred_tree, y_prob_tree = train_and_predict('tree', X_train, y_train, X_val, y_val, X_test)

metric_tree = evaluate_model(y_test, y_pred_tree, y_prob_tree, model_name='Decision Tree')
results.append(metric_tree)
prob_dict['Decision Tree'] = y_prob_tree
```

```python
# Review Baselines
print_results(results)

# Save baseline models & probabilities for later comparison
os.makedirs('../models', exist_ok=True)
os.makedirs('../outputs', exist_ok=True)

with open('../models/prob_dict_baseline.pkl', 'wb') as f:
    pickle.dump(prob_dict, f)
with open('../outputs/results_baseline.pkl', 'wb') as f:
    pickle.dump(results, f)
```
